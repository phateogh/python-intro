```python
# PRACTICE QUESTION — Topic 8
# Write your inventory system here
#inventory list of products

inventory = [
    ("rice", 4500, 35),
    ("bean", 5000, 25),
    ("yam" , 2500, 15),
    ("cassava", 1000, 100)
]
inventory.append( ("banana", 3500, 19) )
inventory.append( ("apple" , 500,100) )
print(inventory)

#remove selected product by unpacking
for product in inventory:
    name, price, qty = product
    if product[0] == "cassava":
        inventory.remove(product)
print(inventory)
    
#print enumerated inventory
for num, product in enumerate(inventory, start=1):
    print(f"{num} - {product}")

#get total inventory amount    
total_price = 0
for product in inventory:
    name, price, qty = product
    total_price += product[1] * product[2]
print(f"total inventory price = {total_price}")

#get most expensive product
most_expensive = inventory[4]
for product in inventory:
    if product[1] > most_expensive[1]:
        most_expensive = product
print(most_expensive[0:-1]) 
```
